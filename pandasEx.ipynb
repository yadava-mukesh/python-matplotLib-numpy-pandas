**Combing Dataframes**

```python
df1=pd.DataFrame({'City':['Mumbai','Delhi','Banglore','Hydrabad'], 'Temperature':[32,45,40,36]})
df1
```

```python
df2=pd.DataFrame({'City':['Delhi','Mumbai','Banglore','Chennai'], 'Humidity':[68,65,75,70]})
df2
```

**Concatanation**

```python
df3=pd.concat([df1,df2], axis=1)
```

```python
df3
```

```python
df4=pd.concat([df1,df2],axis=0, ignore_index=True)
df4
```

**Merging of Dataframes**

```python
pd.merge(df1,df2, on='City', how='outer')
```
