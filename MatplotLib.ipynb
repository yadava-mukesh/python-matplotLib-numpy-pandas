# Histogram

x axis = variables, Y axis= frequency of variable

##### Important parameter in Histogram 
- 1. data
- 2. bin(default=10)

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['Grade'])
# plt.show()
```

#### Type of Analysis
1. Univariate
2. Bivariate
3. Multivariate

```python
plt.hist(df['Grade'], bins=10, range=(0,200))
plt.xticks(list(range(0,200,20)))
plt.xlabel('Grade')
plt.ylabel('How Many Students')
plt.title('Frequency Distribution of Grade')
plt.show()
```

```python
df['Grade'].skew()
```

#### Box Plot
to check given data has outliers or not

```python
plt.boxplot(df['Grade'])
plt.show()
```

```python
Q3=df['Grade'].quantile(0.75)
Q1=df['Grade'].quantile(0.25)
IQR=Q3-Q1
print(Q3, Q1, IQR)
```

```python
upper_limit=Q3+(IQR*1.5)
lower_limit=Q1-(IQR*1.5)
df[(df['Grade']>upper_limit) | (df['Grade']<lower_limit)]
```

# Stacked bar plot

```python
company = ['google','amazon','microsoft','facebook']
turnover = [90,136,89,27]
profit = [40,46,34,12]
```

```python
plt.bar(company,turnover, label='turnover')
plt.bar(company,profit, label='profit')
plt.legend()
plt.show()
```

# Unstacked bar plot

```python
company = ['google','amazon','microsoft','facebook']
turnover = [90,136,89,27]
profit = [40,46,34,12]

plt.bar([0,1,2,3], turnover, width = 0.4, label='turnover')
plt.bar([0.4,1.4,2.4,3.4], profit, width = 0.4, label='profit')
plt.xticks([0,1,2,3])
plt.show()
```
